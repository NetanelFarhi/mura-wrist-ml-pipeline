# 04 Fine tune ResNet 18

Fit the planned ResNet-18 configuration. Early stopping, aggregation selection, and operating-threshold selection use validation only. Test predictions are written after those choices freeze; final test metrics are deferred to notebook 05.


```python
from pathlib import Path
import os
import subprocess
import sys

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_ROOT = os.environ.get("MURA_DATA_ROOT")
if not DATA_ROOT:
    raise RuntimeError("Set MURA_DATA_ROOT first.")

subprocess.run(
    [
        sys.executable, "scripts/train_resnet18.py",
        "--manifest", "data/manifests/wrist_manifest.csv",
        "--split-manifest", "data/manifests/split_manifest.csv",
        "--output-dir", "models/resnet18",
        "--dataset-root", DATA_ROOT,
        "--allow-weight-download",
        "--write-test-predictions",
    ],
    cwd=ROOT,
    check=True,
)
```
