# 05 Sealed Phase 1 evaluation

Run only after both models' checkpoints, aggregation choices, and operating thresholds are frozen using validation data. This is the one final test-metric computation: it produces the common study-level comparison and error-analysis reports.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
subprocess.run(
    [
        sys.executable, "scripts/evaluate_phase1.py",
        "--evaluation-config", "configs/evaluation_models.json",
        "--manifest", "data/manifests/wrist_manifest.csv",
        "--output-dir", "reports/metrics",
        "--error-output-dir", "reports/errors",
        "--baseline-report", "reports/baseline_results.md",
        "--error-report", "reports/error_analysis.md",
    ],
    cwd=ROOT,
    check=True,
)
```
